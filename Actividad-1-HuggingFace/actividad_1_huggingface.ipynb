# Actividad 1. Obtención y preparación de datos desde Hugging Face

En esta actividad se utiliza el dataset Iris, disponible en Hugging Face, para trabajar con datos mediante Python y pandas.

El proceso consiste en cargar la información, revisar su estructura, aplicar cuatro transformaciones y generar un archivo de texto con el resumen de los datos procesados.

```python
from hf_dataset_loader import HFDatasetLoader
from dataset_summarizer import DatasetSummarizer

import pandas as pd
```

```python
loader = HFDatasetLoader("scikit-learn/iris", split="train")
df_original = loader.cargar_datos()

df_original.head()
```

```python
print("Dimensiones del dataset:", df_original.shape)

print("\nTipos de datos:")
print(df_original.dtypes)

print("\nValores nulos:")
print(df_original.isnull().sum())
```

### Transformación 1. Renombrar columnas

Se cambian los nombres de las columnas para facilitar la lectura y el manejo de los datos. Se conservan los valores originales del dataset.

```python

df_transformado = df_original.rename(columns={
    "Id": "id",
    "SepalLengthCm": "largo_sepalo",
    "SepalWidthCm": "ancho_sepalo",
    "PetalLengthCm": "largo_petalo",
    "PetalWidthCm": "ancho_petalo",
    "Species": "especie"
})

df_transformado.head()
```

### Transformación 2. Filtrar registros

Se seleccionan las flores que tienen una longitud de pétalo mayor a 2 cm. Con este filtro se reduce la cantidad de registros y se conservan únicamente los que cumplen la condición.

```python

df_transformado = df_transformado[
    df_transformado["largo_petalo"] > 2
].copy()

print("Registros después del filtro:", len(df_transformado))

df_transformado.head()
```

### Transformación 3. Crear una columna calculada

Se agrega una nueva columna llamada `area_petalo`, calculada al multiplicar el largo por el ancho del pétalo. Esto permite obtener una medida aproximada de su superficie.

```python

df_transformado["area_petalo"] = (
    df_transformado["largo_petalo"] * df_transformado["ancho_petalo"]
).round(2)

print("Dimensiones del DataFrame:", df_transformado.shape)

df_transformado.head()
```

### Transformación 4: Ordenamiento de registros

Se organizan los registros de mayor a menor según
el largo del pétalo y se reinicia el índice del DataFrame.

```python
df_transformado = df_transformado.sort_values(
    by="largo_petalo",
    ascending=False
).reset_index(drop=True)

print("Dimensiones finales:", df_transformado.shape)
df_transformado.head(10)
```

```python
resumidor = DatasetSummarizer(df_transformado, "Iris transformado")

ruta_resumen = resumidor.guardar_resumen("resumen_dataset.txt")

print(resumidor.generar_resumen())
```
