```python
from datasets import load_dataset
import pandas as pd

# Identificador del dataset en Hugging Face
dataset_id = "dev02chandan/sales-forecast-dataset"

# Consultar la estructura del dataset
dataset = load_dataset(dataset_id)

print(dataset)
```

```python
# Convertir el dataset de Hugging Face a DataFrame
df = dataset["train"].to_pandas()

# Mostrar las primeras cinco filas
display(df.head())

# Mostrar las dimensiones
print("Cantidad de filas:", df.shape[0])
print("Cantidad de columnas:", df.shape[1])

# Mostrar nombres y tipos de columnas
print("\nColumnas y tipos de datos:")
print(df.dtypes)

# Revisar valores nulos
print("\nValores nulos por columna:")
print(df.isnull().sum())
```

```python
from pathlib import Path

# Identificar la carpeta del proyecto
proyecto = Path.cwd()

if proyecto.name == "notebooks":
    proyecto = proyecto.parent

# Crear la carpeta de resultados
carpeta_resultados = proyecto / "resultados"
carpeta_resultados.mkdir(parents=True, exist_ok=True)

# Definir la ruta del archivo
archivo_resumen = carpeta_resultados / "resumen_dataframe.txt"

# Construir el resumen del DataFrame
resumen = [
    "PROYECTO INTEGRADOR EA2 - ANÁLISIS DE VENTAS",
    "=" * 55,
    f"Dataset: {dataset_id}",
    "Split utilizado: train",
    "Muestra utilizada: No. Se utiliza el split train completo.",
    f"Cantidad de filas: {df.shape[0]}",
    f"Cantidad de columnas: {df.shape[1]}",
    "",
    "COLUMNAS, TIPOS DE DATOS Y VALORES NULOS",
    "-" * 55
]

for columna in df.columns:
    resumen.append(
        f"{columna} | Tipo: {df[columna].dtype} | "
        f"Nulos: {df[columna].isnull().sum()}"
    )

# Guardar el archivo en UTF-8
archivo_resumen.write_text("\n".join(resumen), encoding="utf-8")

print("Resumen generado correctamente:")
print(archivo_resumen)

# Verificar el contenido
print("\nCONTENIDO DEL RESUMEN:")
print(archivo_resumen.read_text(encoding="utf-8"))
```
