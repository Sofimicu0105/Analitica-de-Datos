# Actividad 2: Manejo de archivos CSV, JSON y TXT

En esta actividad se implementan clases en Python para importar,
modificar y guardar archivos CSV y JSON, además de generar un
resumen de los resultados en un archivo TXT.

El desarrollo utiliza programación orientada a objetos y un
notebook de Jupyter para ejecutar y verificar el proceso.

```python
from pathlib import Path

from manejador_csv import ManejadorCSV
from manejador_json import ManejadorJSON
from generador_resumen_txt import GeneradorResumenTXT

# Carpeta de trabajo
RUTA_BASE = Path.cwd()

# Rutas de los archivos originales
RUTA_CSV = RUTA_BASE / "datos_originales" / "dev.csv"
RUTA_JSON = RUTA_BASE / "datos_originales" / "tts_voice_design_dataset.json"

# Carpeta donde guardaremos los resultados
RUTA_SALIDA = RUTA_BASE / "datos_modificados"

print("Carpeta de trabajo:", RUTA_BASE)
print("CSV encontrado:", RUTA_CSV.exists())
print("JSON encontrado:", RUTA_JSON.exists())
```

```python
# Crear el objeto para manejar el archivo CSV
manejador_csv = ManejadorCSV(RUTA_CSV)

# Importar los registros
registros_csv = manejador_csv.importar_csv()

# Mostrar el valor original
print("Valor original:", registros_csv[0][manejador_csv.encabezados.index("News Finding")])

# Modificar el primer registro
cambio_csv = manejador_csv.modificar_registro(
    indice=0,
    columna="News Finding",
    nuevo_valor="Registro actualizado para la Actividad 2"
)

# Mostrar la modificación
print("Modificación realizada:", cambio_csv)

# Guardar el archivo modificado
ruta_csv_modificado = RUTA_SALIDA / "dev_modificado.csv"
manejador_csv.guardar_csv(ruta_csv_modificado)

print("Total de registros procesados:", len(manejador_csv.registros))
```

```python
# Crear el objeto para manejar el archivo JSON
manejador_json = ManejadorJSON(RUTA_JSON)

# Importar los registros
registros_json = manejador_json.importar_json()

# Mostrar el registro original
print("Registro original:")
print(registros_json[0])

# Modificar el campo language del primer registro
cambio_json = manejador_json.modificar_registro(
    indice=0,
    campo="language",
    nuevo_valor="es"
)

# Mostrar la modificación realizada
print("\nModificación realizada:")
print(cambio_json)

# Guardar el JSON modificado
ruta_json_modificado = RUTA_SALIDA / "tts_voice_design_modificado.json"
manejador_json.guardar_json(ruta_json_modificado)

print("\nTotal de registros procesados:", len(manejador_json.registros))
```

```python
# Información del procesamiento del CSV
resultados_csv = {
    "url_origen": "https://huggingface.co/datasets/copenlu/spiced/resolve/main/dev.csv",
    "archivo_original": str(RUTA_CSV),
    "archivo_modificado": str(ruta_csv_modificado),
    "registros_procesados": len(manejador_csv.registros),
    "clase": "ManejadorCSV",
    "metodos": ["importar_csv", "modificar_registro", "guardar_csv"],
    "modificaciones": [
        f"Registro {cambio_csv['indice']}, columna '{cambio_csv['columna']}': "
        f"'{cambio_csv['valor_anterior']}' -> '{cambio_csv['valor_nuevo']}'"
    ]
}

# Información del procesamiento del JSON
resultados_json = {
    "url_origen": "https://huggingface.co/datasets/BreezeBlue/TTS-Voice-Design-Benchmark/resolve/main/tts_voice_design_dataset.jsonl",
    "archivo_original": str(RUTA_JSON),
    "archivo_modificado": str(ruta_json_modificado),
    "registros_procesados": len(manejador_json.registros),
    "clase": "ManejadorJSON",
    "metodos": ["importar_json", "modificar_registro", "guardar_json"],
    "modificaciones": [
        f"Registro {cambio_json['indice']}, campo '{cambio_json['campo']}': "
        f"'{cambio_json['valor_anterior']}' -> '{cambio_json['valor_nuevo']}'"
    ]
}

# Generar y guardar el resumen
generador = GeneradorResumenTXT(resultados_csv, resultados_json)

ruta_resumen = RUTA_BASE / "resumen.txt"
generador.guardar_resumen(ruta_resumen)

print("Resumen generado en:", ruta_resumen)
print("\nCONTENIDO DEL RESUMEN:\n")
print(ruta_resumen.read_text(encoding="utf-8"))
```

```python
import csv
import json

# Verificar CSV
with open(RUTA_CSV, "r", encoding="utf-8-sig", newline="") as archivo:
    lector = csv.DictReader(archivo)
    csv_original = next(lector)

with open(ruta_csv_modificado, "r", encoding="utf-8-sig", newline="") as archivo:
    lector = csv.DictReader(archivo)
    csv_modificado = next(lector)

# Verificar JSON
with open(RUTA_JSON, "r", encoding="utf-8") as archivo:
    json_original = json.load(archivo)

with open(ruta_json_modificado, "r", encoding="utf-8") as archivo:
    json_modificado = json.load(archivo)

# Comprobar que los originales se conservaron
assert csv_original["News Finding"] != csv_modificado["News Finding"]
assert json_original[0]["language"] == "en"
assert json_modificado[0]["language"] == "es"
assert len(json_original) == len(json_modificado) == 1000

print("VERIFICACIÓN EXITOSA")
print("CSV: archivo original conservado y modificación guardada.")
print("JSON: archivo original conservado y modificación guardada.")
print("Total de registros JSON:", len(json_original))
```
